```python
# clean a list of lines
def clean_pairs(lines):
    cleaned = list()
    # prepare regex for char filtering
    re_print = re.compile('[^%s]' % re.escape(string.printable))
    # prepare translation table for removing punctuation
    table = str.maketrans('', '', string.punctuation)

    for pair in lines:
        clean_pair = list()
        for line in pair:
            # normalize unicode characters
            line = normalize('NFD', line).encode('ascii', 'ignore')
            line = line.decode('utf-8')
            # tokenize on white space
            line = line.split()
            # convert to lowercase
            line = [word.lower() for word in line]
            # remove punctuation from each token
            line = [word.translate(table) for word in line]
            # remove non-printable chars from each token
            line = [re_print.sub('', w) for w in line]
            # remove tokens with numbers in them
            line = [word for word in line if word.isalpha()]
            # store as string
            clean_pair.append(' '.join(line))
        cleaned.append(clean_pair)
    return array(cleaned)

# save a list of clean sentences to file
def save_clean_data(sentences, filename):
    dump(sentences, open(filename, 'wb'))
    print('Saved: %s' % filename)
```

```python
# load dataset
filename = '/content/fra.txt'
doc = load_doc(filename)

# split into english-french pairs
raw_pairs = to_pairs(doc) # Store the original output from to_pairs

# Filter raw_pairs to keep only the first two elements for each pair
filtered_pairs = []
for pair in raw_pairs:
    # Ensure each pair has at least two elements, taking only the first two
    if len(pair) >= 2:
        filtered_pairs.append([pair[0], pair[1]])
    else:
        # Handle cases where a line might have fewer than 2 elements
        # This ensures all filtered_pairs are of length 2
        processed_pair = ['', '']
        if len(pair) > 0:
            processed_pair[0] = pair[0]
        filtered_pairs.append(processed_pair)

# clean sentences using the now consistently-sized pairs
cleaned_sentence_pairs = clean_pairs(filtered_pairs)

# save clean pairs to file
save_clean_data(cleaned_sentence_pairs, 'english-french.pkl')

# spot check
for i in range(100):
    print('[%s] => [%s]' % (cleaned_sentence_pairs[i,0], cleaned_sentence_pairs[i,1]))
```

```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from keras.preprocessing.sequence import pad_sequences
from keras.utils import to_categorical
from keras.models import Sequential
from keras.layers import LSTM, Dense, Embedding, RepeatVector, TimeDistributed
from keras.callbacks import ModelCheckpoint

# --- 1. REDEFINE FUNCTIONS (The Fix) ---
def create_tokenizer(lines):
    tokenizer = Tokenizer()
    tokenizer.fit_on_texts(lines)
    return tokenizer

def max_length(lines):
    return max(len(line.split()) for line in lines)

def encode_sequences(tokenizer, length, lines):
    X = tokenizer.texts_to_sequences(lines)
    X = pad_sequences(X, maxlen=length, padding='post')
    return X

def encode_output(sequences, vocab_size):
    ylist = list()
    for sequence in sequences:
        encoded = to_categorical(sequence, num_classes=vocab_size)
        ylist.append(encoded)
    y = np.array(ylist)
    y = y.reshape(sequences.shape[0], sequences.shape[1], vocab_size)
    return y

def define_model(src_vocab, tar_vocab, src_timesteps, tar_timesteps, n_units):
    model = Sequential()
    model.add(Embedding(src_vocab, n_units, input_length=src_timesteps, mask_zero=True))
    model.add(LSTM(n_units))
    model.add(RepeatVector(tar_timesteps))
    model.add(LSTM(n_units, return_sequences=True))
    model.add(TimeDistributed(Dense(tar_vocab, activation='softmax')))
    return model

# --- 2. DATA PREPARATION ---
# (Make sure 'dataset', 'train', and 'test' variables exist from your previous steps)

eng_tokenizer = create_tokenizer(dataset[:, 0])
eng_vocab_size = len(eng_tokenizer.word_index) + 1
eng_length = max_length(dataset[:, 0])

fra_tokenizer = create_tokenizer(dataset[:, 1])
fra_vocab_size = len(fra_tokenizer.word_index) + 1
fra_length = max_length(dataset[:, 1])

trainX = encode_sequences(eng_tokenizer, eng_length, train[:, 0])
trainY = encode_sequences(fra_tokenizer, fra_length, train[:, 1])
trainY = encode_output(trainY, fra_vocab_size)

testX = encode_sequences(eng_tokenizer, eng_length, test[:, 0])
testY = encode_sequences(fra_tokenizer, fra_length, test[:, 1])
testY = encode_output(testY, fra_vocab_size)

# --- 3. MODEL TRAINING ---
model = define_model(eng_vocab_size, fra_vocab_size, eng_length, fra_length, 256)
model.compile(optimizer='adam', loss='categorical_crossentropy')

checkpoint = ModelCheckpoint('model.keras', monitor='val_loss', save_best_only=True, mode='min')
model.fit(trainX, trainY, epochs=30, batch_size=64, validation_data=(testX, testY), callbacks=[checkpoint])
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from keras.preprocessing.sequence import pad_sequences
from keras.models import load_model
from pickle import load

# 1. FINAL TRANSLATION FUNCTION
def translate_manual(sentence, model, eng_tokenizer, fra_tokenizer, eng_length):
    # Convert text to integers
    seq = eng_tokenizer.texts_to_sequences([sentence])
    # Pad to match the model's expected input length
    padded = pad_sequences(seq, maxlen=eng_length, padding='post')

    # Predict using direct call (fixes the batch_outputs bug)
    source = np.array(padded, dtype='int32')
    prediction = model(source, training=False)

    # Map numbers back to French words
    integers = [np.argmax(vector) for vector in prediction.numpy()[0]]
    target = []
    for val in integers:
        word = word_for_id(val, fra_tokenizer)
        if word is None: break
        target.append(word)
    return " ".join(target)

# 2. RE-INITIALIZE AND TEST
try:
    # Reloading to ensure everything is fresh
    dataset = load(open('english-french-both.pkl', 'rb'))
    model = load_model('model.keras')

    eng_tokenizer = create_tokenizer(dataset[:, 0])
    fra_tokenizer = create_tokenizer(dataset[:, 1])
    eng_length = max_length(dataset[:, 0])

    print("✅ Initialization Complete!")
    print("-" * 40)

    # TEST WITH YOUR OWN SENTENCES
    test_sentences = ["i love you", "it is cold", "he is happy", "she is fast"]

    for text in test_sentences:
        result = translate_manual(text, model, eng_tokenizer, fra_tokenizer, eng_length)
        print(f"English: {text}")
        print(f"French:  {result}")
        print("-" * 40)

except Exception as e:
    print(f"❌ Execution Error: {e}")
```

```python
from nltk.translate.bleu_score import corpus_bleu, SmoothingFunction

def evaluate_model(model, tokenizer, sources, raw_dataset):
    actual, predicted = list(), list()
    # 1. Initialize Smoothing to prevent the ZeroDivisionError
    smoother = SmoothingFunction().method1

    print("Evaluating model quality... Please wait.")

    for i, source in enumerate(sources):
        # Prepare input shape
        source_input = source.reshape((1, source.shape[0]))

        # Get prediction (Direct call to bypass memory bugs)
        prediction = model(source_input, training=False)
        integers = [np.argmax(vector) for vector in prediction.numpy()[0]]

        # Decode integers to French words
        target_words = []
        for val in integers:
            word = word_for_id(val, tokenizer)
            if word is None: break
            target_words.append(word)

        # Get original French sentence
        raw_target = raw_dataset[i, 1]
        actual.append([raw_target.split()])
        predicted.append(target_words)

    # 2. Calculate scores using the smoothing_function
    print("-" * 40)
    b1 = corpus_bleu(actual, predicted, weights=(1.0, 0, 0, 0), smoothing_function=smoother)
    b2 = corpus_bleu(actual, predicted, weights=(0.5, 0.5, 0, 0), smoothing_function=smoother)
    b3 = corpus_bleu(actual, predicted, weights=(0.33, 0.33, 0.33, 0), smoothing_function=smoother)
    b4 = corpus_bleu(actual, predicted, weights=(0.25, 0.25, 0.25, 0.25), smoothing_function=smoother)

    print(f'BLEU-1 (Individual Words): {b1:.4f}')
    print(f'BLEU-2 (Two-word phrases): {b2:.4f}')
    print(f'BLEU-3 (Three-word phrases): {b3:.4f}')
    print(f'BLEU-4 (Full sentence flow): {b4:.4f}')

# Run the final evaluation
evaluate_model(model, fra_tokenizer, testX[:100], test[:100])
```
